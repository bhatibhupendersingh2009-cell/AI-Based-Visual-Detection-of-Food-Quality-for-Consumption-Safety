```python
# Load necessary Libraries.

from tensorflow.keras.models import load_model

from tensorflow.keras.preprocessing import image

import matplotlib.pyplot as plt

import numpy as np
```

```python
# Load the trained model.

model = load_model("food_quality_model.keras")
```

```python
# Load Image
img = image.load_img(

    r"C:\AI-Based Visual Detection of Food Quality for Consumption Safety\Test_Crispy_Toast_5.jpg",

    target_size=(224,224)
)

# Show Image
plt.imshow(img)

plt.axis("off")

# Convert To Array
img_array = image.img_to_array(img)

# Expand Dimensions
img_array = np.expand_dims(

    img_array,

    axis=0
)

# Normalize
img_array = img_array / 255

# Prediction
prediction = model.predict(img_array)

prediction_value = prediction[0][0]

# Final Prediction Logic
if prediction_value >= 0.5:

    result = "Spoiled Food"

    confidence = prediction_value * 100

else:

    result = "Fresh Food"

    confidence = (1 - prediction_value) * 100

# Show Result
plt.title(

    f"{result}\nConfidence : {round(confidence,2)}%"
)

plt.show()

# Print Result
print("Food Status :", result)

print("Confidence :", round(confidence,2), "%")
```
